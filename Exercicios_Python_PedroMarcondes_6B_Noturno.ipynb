```python
# Ex 1 - Hello, World!
print("hello, world!")

# Ex 2 - Soma de dois números
print("por favor, digite dois números para realizarmos uma soma")

num1 = int(input("primeiro número: "))
num2 = int(input("segundo número: "))

print("a soma de", num1, "mais", num2, "é igual a:", num1 + num2)
print(f"a soma de {num1} mais {num2} é igual a: {num1 + num2}") ## professor, coloquei duas versões, uma para lembrar do uso da concatenação

# Ex 3 - Cálculo da Área do Círculo
import math ## eu desenvolvo profissionalmente usando C#, por padrão vem com using system, então admito que quase fugiu da cabeça o import do math

print("por favor, digite a seguir o raio de um círculo para calcularmos a área")

raio = float(input())

area = math.pi * (raio ** 2)

print("a área do círculo é:", area)
print(f"a área do círculo é: {area:.2f}") ## professor, coloquei duas versões mais uma vez, agora usando formatação de casas decimais

# Ex 4 - Cálculo da Área do Círculo
print("por favor, digite uma temperatura em celsius para convertermos para fahrenheit:")

celsius = float(input())

fahrenheit = celsius * 9 / 5 + 32

print(f"a temperatura em fahrenheit é: {fahrenheit:.2f}")

# Ex 5 - Tabuada
print("por favor, digite um número para vermos a tabuada dele")

num = int(input())

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")


# Ex 6 - Números pares

print("a seguir, todos os números pares de 1 a 20")
for numero in range(1, 21):
    if numero % 2 == 0:
        print(numero)

# Ex 7 - Números Ímpar ou Par
print("por favor, digite um número para descobrirmos se é ímpar ou par")

num = int(input())

if num % 2 == 0:
 print("é par!")
else:
 print("é impar!")

# Ex 8 - Verificação de Palíndromo
print("por favor, digite uma palavra para descobrirmos se essa palavra é um palíndromo")

palavra = input()

if palavra == palavra[::-1]: ## aqui eu tive que pesquisar pra lembrar, inicialmente pensei em um .reverse mas não é uma lista
  print ("opa, sua palavra é um palíndromo!")
else:
  print("não é um palíndromo!")

# Ex 9 - Fatorial de um número
print("por favor, digite um número")

num = int(input())

fatorial = 1

for i in range(1, num + 1):
    fatorial = fatorial * i

print(f"o fatorial é: {fatorial}")

# Ex 10 - Sequência de Fibonacci
a = 0
b = 1

for i in range(10):
    print(a)

    c = a + b
    a = b
    b = c

# Ex 11 - Lista de Números
lista = [5, 12, 3, 67, 7, 127]
print(f"o mario número é: {max(lista)}")
print(f"o menor número é: {min(lista)}")

# Ex 12 - Contagem de caracteres
print("por favor, digite uma palavra: ")

palavra = input()

count = {}

for carac in palavra:
    if carac in count:
        count[carac] = count[carac] + 1 ## professor, coloquei pra não ter repetição na contagem dos caracteres
    else:
        count[carac] = 1

print(f"essa foi a quantidade de caracteres únicos {count}")

# Ex 13 - Ordenação de lista
listinha = [3, 21, 17, 67 , 2 , 9]
listinha.sort()
print(f"lista ordenada: {listinha}")

# Ex 14 - Soma dos elementos de uma pista
listinha = [34, 21, 3, 8 , 73, 127]
soma = sum(listinha)
print(f"lista somada: {soma}")

# Ex 15 - Média de uma Lista
listinha = [12, 31, 78, 23, 1, 54, 9]
soma = sum(listinha)
quant = len(listinha)
media = soma / quant
print(f"a média é: {media}")

# Ex 16 - Manipulação de Strings
print("por favor, digite uma palavra: ")

palavra = input()
invert = palavra[::-1]

print(f"veja a sua palavra invertida {invert}")

# Ex 17 - Dicionário de Contagem de Palavras
print("por favor, digite uma frase do tamanho que quiser:")

frase = input()

palavras = frase.split()

contagem = {}

for palavra in palavras:
    if palavra in contagem:
        contagem[palavra] = contagem[palavra] + 1
    else:
        contagem[palavra] = 1

print(f"sua frase possui essa quantidade de palavras: {contagem}")

# Ex 18 - Número Primo
num = int(input("por favor, digite um número: ")) # professor, estou acostumado a usar um print com a mensagem e associar o output do usuário na váriavel seguinte. estive olhando aulas de python e fazem dessa maneira do exercício, mas continuarei fazendo do meu jeito após esse ok

primo = True

if num < 2:
    primo = False
else:
    for i in range(2, num):
        if num % i == 0:
            primo = False
            break

if primo:
    print("é primo!")
else:
    print("não é primo!")

# Ex 19 - Jogo de Adivinhação
import random

secreto = random.randint(1, 100)

tentativa = 0

while tentativa != secreto:
    print("tente adivinhar o número entre 1 e 100, boa sorte! ")
    tentativa = int(input())

    if tentativa < secreto:
        print("mais alto!")
    elif tentativa > secreto:
        print("mais baixo!")

print("você acertou (finalmente)!")

# Ex 20 - Matriz Transposta
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposta = []

for i in range(3):
    linha = []

    for j in range(3):
        linha.append(matriz[j][i])

    transposta.append(linha)

print(transposta)

# Ex 21 - Busca Binária
# Professor, uma dúvida. Python recomenda-se usar snake_case, mas estou acostumado com camelCase, tem problema?
def busca_binaria(lista, alvo):

    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:

        meio = (inicio + fim) // 2

        if lista[meio] == alvo:
            return meio

        elif lista[meio] < alvo:
            inicio = meio + 1

        else:
            fim = meio - 1

    return -1


lista_ordenada = [1, 3, 5, 7, 9, 11, 13]

resultado = busca_binaria(lista_ordenada, 7)

print(resultado)

# Ex 22 - Criptografia de Cifra de César
def criptografar_cesar(texto, deslocamento):

    resultado = ""

    for letra in texto:

        if letra.isalpha(): ## aqui eu tive que pesquisar, não me recordava do isalpha

            codigo = ord(letra)

            novo_codigo = codigo + deslocamento

            resultado = resultado + chr(novo_codigo)

        else:
            resultado = resultado + letra

    return resultado


texto = "Aluno Pedro Marcondes!"

resultado = criptografar_cesar(texto, 3)

print(resultado)

# Ex 23 - Descriptografar Cifra de César
def descriptografar_cesar(texto, deslocamento):

    resultado = ""

    for letra in texto:

        if letra.isalpha():

            codigo = ord(letra)

            novo_codigo = codigo - deslocamento

            resultado = resultado + chr(novo_codigo)

        else:
            resultado = resultado + letra

    return resultado


texto = "Doxqr S hg urpdqghv!"

resultado = descriptografar_cesar(texto, 3)

print("textoo criptografado:", texto)
print("texto original:", resultado)

# Ex 24 - Validação de CPF
def validar_cpf(cpf):

    cpf = cpf.replace(".", "")
    cpf = cpf.replace("-", "")

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0

    for i in range(9):
        soma = soma + int(cpf[i]) * (10 - i)

    resto = (soma * 10) % 11

    if resto == 10:
        resto = 0

    if resto != int(cpf[9]):
        return False

    soma = 0

    for i in range(10):
        soma = soma + int(cpf[i]) * (11 - i)

    resto = (soma * 10) % 11

    if resto == 10:
        resto = 0

    if resto != int(cpf[10]):
        return False

    return True


cpf = input("por favor, digite seu cpf: ")

if validar_cpf(cpf):
    print("cpf válido!")
else:
    print("opa, cpf inválido!")

# Ex 25 - Classificação de Números
lista = [5, -3, 0, 12, -8, 0, 7]

positivos = []
negativos = []
zeros = []

for numero in lista:

    if numero > 0:
        positivos.append(numero)

    elif numero < 0:
        negativos.append(numero)

    else:
        zeros.append(numero)

print(f"positivos: {positivos}")
print(f"negativos: {negativos}")
print(f"zeros: {zeros}")

# Ex 26 - Calculadora Simples
print("vamos usar uma calculadora. primeiramente, digite o primeiro número: ")

num1 = float(input())

print("agora, o operador (+, -, *, /): ")

operador = input()

print("e por último, o segundo número: ")

num2 = float(input())

if operador == "+":
    print(num1 + num2)
elif operador == "-":
    print(num1 - num2)
elif operador == "*":
    print(num1 * num2)
elif operador == "/":
    if num2 == 0:
        print("opa, divisão por zero não tem como")
    else:
        print(num1 / num2)

# Ex 27- Anagramas
print("vamos realizar um jogo para descobrirmos se duas palavras são anagramas!")
print("por favor, digite a primeira palavra: ")

pal1 = input()

print("por favor, digite a segunda palavra: ")

pal2 = input()

if sorted(pal1.lower()) == sorted(pal2.lower()): # demorei um pouco pra lembrar do sorted pra essa situação
    print("São anagramas!")
else:
    print("Não são anagramas!")

# Ex 28 - Gráfico de Barras
import matplotlib.pyplot as plt # esse exercício eu pesquisei como realizar, não havia trabalhado em cima dessa lib ainda

categorias = []

valores = []

print("vamos montar um gráfico de barras!")

for i in range(4):
    categ = input("digite a categoria: ")
    valor = float(input("digite o valor: "))

    categorias.append(categ)
    valores.append(valor)

plt.bar(categorias, valores)
plt.show()

# Ex 29 - Números FizzBuzz
for f in range(1, 101):
    if f % 15 == 0:
        print("FizzBuzz")
    elif f % 3 == 0:
        print("Fizz")
    elif f % 5 == 0:
        print("Buzz")
    else:
        print(f)

# Ex 30 - jogo da forca
palavra = "marcondes"
descobertas = ["_"] * len(palavra)
tentativas = 5

while tentativas > 0 and "_" in descobertas:

    print("\n" + " ".join(descobertas))

    letra = input("digite uma letra: ").lower()

    if letra in palavra:

        for i, l in enumerate(palavra):
            if l == letra:
                descobertas[i] = letra

        print("acertou!")

    else:
        tentativas -= 1
        print(f"errou! tentativas restantes: {tentativas}")

if "_" not in descobertas:
    print(f"você venceu! a palavra era: {palavra}")
else:
    print(f"você perdeu! a palavra era: {palavra}")

# Ex 31 - Ordenação por Bolha
def ordenar(lista):
    n = len(lista)

    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista

print(ordenar([5, 12, 3, 47, 8, 21]))


# ex 32 - jogo da velha
tabuleiro = [" "] * 9

def imprimir():
    for i in range(0, 9, 3):
        print(tabuleiro[i:i + 3])


def vencedor():
    combinacoes = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in combinacoes:
        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
            return tabuleiro[a]

    return None

jogador = "x" # essa foi a melhor maneira de definir dois jogadores? foi a que eu achei ideal pesquisando

while " " in tabuleiro and not vencedor():
    imprimir()

    posicao = int(input(f"jogador {jogador}, escolha uma posição (0-8): "))

    if tabuleiro[posicao] == " ":
        tabuleiro[posicao] = jogador

        if jogador == "x":
            jogador = "o"
        else:
            jogador = "x"

vencedor = vencedor()

imprimir()

if vencedor:
    print(f"vencedor: {vencedor}")
else:
    print("empate!")

# Ex 33 - verificação de anagramas - opa professor, exercício repetido
print("vamos realizar um jogo para descobrirmos se duas palavras são anagramas!")
print("por favor, digite a primeira palavra: ")

pal1 = input()

print("por favor, digite a segunda palavra: ")

pal2 = input()

if sorted(pal1.lower()) == sorted(pal2.lower()):
    print("São anagramas!")
else:
    print("Não são anagramas!")


# Ex 34 - Contador de Vogais
print("vamos contar quantas vogais existem em um texto!")
print("por favor, digite um texto: ")

texto = input()

vogais = "aeiou"
contador = 0

for letra in texto.lower():
    if letra in vogais:
        contador += 1

print(f"o texto possui {contador} vogais!")

# Ex 35 - Números fibonacci até n
print("vamos mostrar os números da sequência fibonacci!")
print("por favor, digite um número para ser o limite: ")

n = int(input())

a = 0
b = 1

while a <= n:
    print(a)
    a, b = b, a + b

# Ex 36 - Gerador de senhas
import random
import string

print("deixa eu gerar uma nova senha para você! digite o tamanho da senha: ")

tamanho = int(input())

caract = string.ascii_letters + string.digits ## pesqquisei soibre essa lib de string, muito interessante
senha = ""

for i in range(tamanho):
    senha += random.choice(caract)

print(f"senha gerada: {senha}")

# ex 37 - Funções Recursivas
def fatorial(num):
    if num == 0 or num == 1:
        return 1

    return num * fatorial(num - 1)


print("vamos calcular o fatorial de um número!")
print("digite um número: ")

num = int(input())

print(f"o fatorial de {num} é {fatorial(num)}")


# ex 38 - contar palavras em texto
print("vamos contar quantas palavras existem em uma frase. por favor, digite um texto: ")

texto = input()

palavras = texto.split()

print(f"o texto possui {len(palavras)} palavras!")


# ex 39 - jogo de dados
import random

print("vamos jogar dois dados!")

dado1 = random.randint(1, 6)
dado2 = random.randint(1, 6)

total = dado1 + dado2

print(f"resultado do primeiro dado: {dado1}")
print(f"resultado do segundo dado: {dado2}")
print(f"total: {total}")

# ex 40 - lista de compras
lista = []

while True:

    print("1 - adicionar item")
    print("2 - remover item")
    print("3 - listar iteens")
    print("4 - sair")

    print("escolha uma opção: ")
    opcao = input()

    if opcao == "1":

        print("digite o item que deseja adicionar: ")
        item = input()

        lista.append(item)

        print("item adicionado!")

    elif opcao == "2":

        print("digite o item que deseja remover: ")
        item = input()

        if item in lista:
            lista.remove(item)
            print("item removido!")
        else:
            print("item não encontrado!")

    elif opcao == "3":

        print("lista de compras:")

        for item in lista:
            print(item)

    elif opcao == "4":

        print("programa encerrado!")
        break

    else:
        print("opção inválida!")

```
