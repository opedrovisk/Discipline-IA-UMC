## Ex 11 - Pandas

**Enunciado:** Preencha os valores nulos de uma coluna numérica com a mediana dessa coluna.

```python
produtos = ['Livro', 'Caderno', 'Caneta', 'Borracha', 'Mochila']
precos = [29.90, 15.50, 3.20, 2.50, 79.90]
quantidade = [10, 20, 50, 30, 5]
categoria = ['Papelaria', 'Papelaria', 'Papelaria', 'Papelaria', 'Acessorios']

dados = {'Produto': produtos, 'Preco': precos, 'Quantidade': quantidade, 'Categoria': categoria}

df = pd.DataFrame(dados)

df['Total'] = df['Preco'] * df['Quantidade']

df.loc[2, 'Preco'] = pd.NA

mediana = df.median(numeric_only=True)
df.fillna(mediana, inplace=True)
print(df)
```
