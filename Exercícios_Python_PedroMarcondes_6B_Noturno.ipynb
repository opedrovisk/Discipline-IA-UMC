## Ex 21 - Busca Binária

```python
# Ex 21 - Busca Binária
# Professor, uma dúvida. Python recomenda-se usar snake_case, mas estou acostumado com camelCase, tem problema?
def busca_binaria(lista, alvo):

    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:

        meio = (inicio + fim) // 2

        if lista[meio] == alvo:
            return meio

        elif lista[meio] < alvo:
            inicio = meio + 1

        else:
            fim = meio - 1

    return -1


lista_ordenada = [1, 3, 5, 7, 9, 11, 13]

resultado = busca_binaria(lista_ordenada, 7)

print(resultado)
```

## Ex 22 - Criptografia de Cifra de César

```python
# Ex 22 - Criptografia de Cifra de César
# pesquisando como usar criptografia no python, já que não me recordava do isalpha por exemplo. usei ord e chr pra mexer no codigo ascii da letra e mover ela
def criptografar_cesar(texto, deslocamento):

    resultado = ""

    for letra in texto:

        if letra.isalpha():

            codigo = ord(letra)

            novo_codigo = codigo + deslocamento

            resultado = resultado + chr(novo_codigo)

        else:
            resultado = resultado + letra

    return resultado


texto = "Aluno Pedro Marcondes!"

resultado = criptografar_cesar(texto, 3)

print(resultado)
```

## Ex 23 - Descriptografar Cifra de César

```python
# Ex 23 - Descriptografar Cifra de César
# mesmo exercício de cima, inverti a lógica em cima do último pra facilitar o desenv
def descriptografar_cesar(texto, deslocamento):

    resultado = ""

    for letra in texto:

        if letra.isalpha():

            codigo = ord(letra)

            novo_codigo = codigo - deslocamento

            resultado = resultado + chr(novo_codigo)

        else:
            resultado = resultado + letra

    return resultado


texto = "Doxqr S hg urpdqghv!"

resultado = descriptografar_cesar(texto, 3)

print("textoo criptografado:", texto)
print("texto original:", resultado)
```

## Ex 24 - Validação de CPF

```python
# Ex 24 - Validação de CPF
# me recordei de ter aprendido isso em uma aula de python do youtube, então segui o cálculo oficial dos dígitos verificadores, só checar o tamanho não garante que o cpf é válido de verdade
def validar_cpf(cpf):

    cpf = cpf.replace(".", "")
    cpf = cpf.replace("-", "")

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0

    for i in range(9):
        soma = soma + int(cpf[i]) * (10 - i)

    resto = (soma * 10) % 11

    if resto == 10:
        resto = 0

    if resto != int(cpf[9]):
        return False

    soma = 0

    for i in range(10):
        soma = soma + int(cpf[i]) * (11 - i)

    resto = (soma * 10) % 11

    if resto == 10:
        resto = 0

    if resto != int(cpf[10]):
        return False

    return True


cpf = input("por favor, digite seu cpf: ")

if validar_cpf(cpf):
    print("cpf válido!")
else:
    print("opa, cpf inválido!")
```

## Ex 31 - Ordenação por Bolha

```python
# Ex 31 - Ordenação por Bolha
# já que aqui pediu o desenv da ordenação, não usei o sort pronto. fiz um for que vai passando pela lista, coloquei em uma função
def ordenar(lista):
    n = len(lista)

    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista

print(ordenar([5, 12, 3, 47, 8, 21]))
```

```python
# ex 32 - jogo da velha
# esse deu uma quebra de cabeça, mas coloquei as combinações vencedoras numa lista de tuplas pra não repetir o mesmo if várias vezes
tabuleiro = [" "] * 9

def imprimir():
    for i in range(0, 9, 3):
        print(tabuleiro[i:i + 3])


def vencedor():
    combinacoes = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in combinacoes:
        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
            return tabuleiro[a]

    return None

jogador = "x" # essa foi a melhor maneira de definir dois jogadores? foi a que eu achei ideal pesquisando

while " " in tabuleiro and not vencedor():
    imprimir()

    posicao = int(input(f"jogador {jogador}, escolha uma posição (0-8): "))

    if tabuleiro[posicao] == " ":
        tabuleiro[posicao] = jogador

        if jogador == "x":
            jogador = "o"
        else:
            jogador = "x"

vencedor = vencedor()

imprimir()

if vencedor:
    print(f"vencedor: {vencedor}")
else:
    print("empate!")
```

## ex 46 - Calcular média e Desvio padrão

```python
# ex 46 - Calcular média e Desvio padrão
# fiz na mão em vez mas vi que dava pra usar uma lib chamada statistics. primeiro a média, depois o quanto cada valor se afasta dela ao quadrado, e no final tiro a raiz
lista = [5, 12, 3, 47, 8, 21]

print("vamos calcular a média e o desvio padrão da lista!")

soma = 0

for numero in lista:
    soma += numero

media = soma / len(lista)

soma = 0

for numero in lista:
    soma += (numero - media) ** 2

desvio = (soma / len(lista)) ** 0.5

print(f"média: {media:.2f}")
print(f"desvio padrão: {desvio:.2f}")
```

```python
# ex 47 - contar números em lista
# mesma lógica de contagem com dicionário que já usei alguns exs atrás
lista = [1, 2, 2, 3, 4, 4, 4, 5]
contagem = {}

print("vamos contar quantas vezes cada número aparece na lista!")

for numero in lista:

    if numero in contagem:
        contagem[numero] += 1
    else:
        contagem[numero] = 1

print(f"quantidade de cada número: {contagem}")
```

## ex 48 - ordenação por inserção

```python
# ex 48 - ordenação por inserção
# o ex pede o algoritmo específico, então não deu pra fugir pro sort
def ordenar(lista):

    for i in range(1, len(lista)):

        chave = lista[i]
        j = i - 1

        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = chave

    return lista


lista = [5, 12, 3, 47, 8, 21]

print("vamos ordenar a lista usando o método de inserção!")

print(f"lista ordenada: {ordenar(lista)}")
```

## ex 49 - soma dos dígitos

```python
# ex 49 - soma dos dígitos
# usei str(abs(num)) pra poder percorrer cada dígito
print("vamos calcular a soma dos dígitos de um número!")
print("por favor, digite um número: ")

num = int(input())

soma = 0

for digito in str(num):
    soma += int(digito)

print(f"a soma dos dígitos é: {soma}")
```
